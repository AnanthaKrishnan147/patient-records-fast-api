```python
def insert_the_second_time(name :str , age:int):
    if type(name) == str and type(age) == int:
        print(name)
        print(age)
        print("value inserted")

    else :
        raise TypeError("Wrong datatype is given as input")

insert_the_second_time('Ananthan' , 'one')
```

the above method is not scalable . doing in this method requies the coding of lot of boilerplate code . 
suppose that we need to do some value checking such as the age should not be negetive . again a lot of effort 

this code will do type validation and data validation .

```python
def insert_the_second_time(name :str , age:int):
    if type(name) == str and type(age) == int:
        if age < 0:
            raise ValueError("Age cannot be negative")
        else :
            print(name)
            print(age)
            print("value inserted")

    else :
        raise TypeError("Wrong datatype is given as input")

insert_the_second_time('Ananthan' , -1)
```
