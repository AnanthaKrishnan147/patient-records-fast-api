```python
def insert_patient(name:str , age:int):
    print(name)
    print(age)
    print("Patient data has been inserted")

insert_patient('ananthan' , 20)
insert_patient('Frank' , 'seventy')
```

# this code will produce error 

```python
def insert_the_second_time(name :str , age:int):
    if type(name) == str and type(age) == int:
        print(name)
        print(age)
        print("value inserted")

    else :
        raise TypeError("Wrong datatype is given as input")

insert_the_second_time('Ananthan' , 'one')
```
